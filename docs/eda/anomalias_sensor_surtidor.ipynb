En los sensores se observa que el registro presenta un repunte a las 08:36:00.000000, con un valor de aproximadamente 2312.63 litros. El registro más cercano corresponde al SST. El algoritmo opera mediante un sistema de puntuación para realizar la correlación de eventos. En este caso, la diferencia temporal con el evento de TruckShop es de 6:29:57, mientras que con SST es de solo 00:36:00. Esta diferencia es considerable, por lo que el algoritmo asocia el registro del evento de recarga con SST y no con TruckShop.

Otro aspecto que también se toma en cuenta es la coincidencia en la cantidad de combustible. En este caso, nuevamente el evento del sensor se asocia más estrechamente con el de SST: la discrepancia entre SST y el sensor es de 47.64 litros, mientras que con TruckShop es de 324.36 litros.

Adicionalmente, el origen del evento también otorga puntuación, ya que un origen TruckShop suele ser más confiable. Sin embargo, incluso con estos puntos extra por origen, no se puede forzar la correlación con el evento de TruckShop debido a las marcadas diferencias en el tiempo y la cantidad de combustible.

El último aspecto considerado es el turno. En este caso, ambos eventos ocurren durante el turno día. Por lo tanto, se concluye que, entre los registros de TruckShop y SST, el de SST es el más coherente con el del sensor, y por ello fue asociado, en base a los criterios de puntuación explicados, el intentar forzar esta correlacion se pierden otras mas correlaciones que realmente son mas cercanas a los eventos de sensor.

Este caso es similar al T-210. El SST es el que más se parece al registro de evento de recarga del sensor. Un detalle es que, ese día, se registran al menos dos eventos en el sensor y tres en el FMS. Al analizar la gráfica y los registros, se observa claramente que podría haber un registro del sensor que se haya omitido. Sin embargo, esto representa un problema considerable, ya que la recarga indicada es de 182 litros, una cantidad demasiado baja para considerarse una recarga real.

Además, esta cantidad es tan baja que no cumple con los criterios de clasificacion como recarga. Las restricciones para detectar recargas: el sensor debe superar un umbral de 190 registros en al menos 30 segundos; el promedio posterior debe ser mayor al promedio anterior más el umbral de 190; y la recarga debe ser de al menos 500 litros.

Este último criterio es crítico debido a que existen ruidos, mesetas y otras anomalías previamente descritas, que generan falsos registros iguales o inferiores a 190. Anteriormente se concluyó que los sensores presentan estas anomalías y que este caso en particular, por tratarse de una recarga tan pequeña, entra en la categoría de anomalía para el algoritmo del sensor. Forzar la inclusión de este evento como un registro de recarga implicaría también aceptar cientos de falsos positivos.
