# CRISP-DM (Cross-Industry Standard Process for Data Mining)

Se usara la Metodología estructurada en **6 fases** para desarrollar proyectos de ciencia de datos de forma ordenada y centrada en resolver problemas del negocio.

## 📌 Fases del CRISP-DM

### 📌 Fase 1: Business Understanding

 📌 **Objetivo:** Desarrollar un modelo predictivo de combustible exclusivamente para camiones en las operaniones de carguio y acarreo en mineria a cielo abierto de MSC


### 📌 Fase 2: Comprensión de los Datos

📌 **Objetivo:** Identificar, recopilar y analizar conjuntos de datos relevantes de nuestro dataset elegido.  


### 📌 Fase 3: Preparación de los Datos

📌 **Objetivo:** Transformar datos en un formato apto para modelado.


### 📌 Fase 4: Modelado

📌 **Objetivo:** Construir y evaluar modelos iterativamente hasta encontrar el óptimo.  

### 📌 Fase 5: Evaluación

📌 **Objetivo:** Validar si el modelo resuelve el problema del negocio y planificar próximos pasos.  

### Fase 6: Despliegue

📌 **Objetivo:** Operacionalizar el modelo para que genere valor en el negocio.
